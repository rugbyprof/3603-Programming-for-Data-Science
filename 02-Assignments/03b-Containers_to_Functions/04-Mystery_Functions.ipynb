# 📘 04 - Mystery Functions

Before you write a function from a blank page, get good at **reading** one. Here's why: when you write from scratch, you have to invent the syntax *and* the logic at the same time, and that's hard. Reading and modifying working code lets you focus on one of them at a time.

## ✅ Learning Goals
- Work out what an unfamiliar function does by **predicting** and **tracing**, before you run it
- Turn one function into several related ones by changing as little as possible
- Diagnose subtly broken functions (the bugs *everyone* writes)
- Know which operations **change** a list passed into a function and which create a new one

## 🔁 The Workflow

For every mystery function in this notebook:

```text
1. Predict  → read the code, then guess the output for a given input
2. Trace    → go through it by hand, writing down every variable's value
3. Run      → check your prediction
4. Explain  → describe what it does in ONE plain-English sentence
5. Modify   → change it to do something slightly different
6. Write    → only now, write one from scratch
```

Don't skip steps 1 and 2. That's where you learn the most.

### Mystery #4 — This One Returns `True` or `False`

Look carefully at where the two `return`s are.

````
def mystery4(values):
    for x in values:
        if x < 0:
            return True
    return False
````

✍️ **`mystery4([5, 3, -1, 8])` → ___   `mystery4([5, 3, 1, 8])` → ___   How many items does the loop look at in the first call?**

````
print(mystery4([5, 3, -1, 8]))
print(mystery4([5, 3, 1, 8]))
````

Remember that `return` **leaves the function immediately**. As soon as `mystery4` finds one negative number, it knows the answer, so it stops checking. It can only answer `False` after checking **everything**, which is why that `return` is outside the loop.

## 🔍 Mutation Detective

When you pass a list into a function, the function gets **the same list**, not a copy. So if the function changes it, the caller's list changes too.

🔮 **Predict** what `original` looks like at the end:

````
def add_bonus(scores):
    for i in range(len(scores)):
        scores[i] += 5
    return scores

original = [70, 80]
curved = add_bonus(original)
print("curved:  ", curved)
print("original:", original)
````

Surprised? `scores` and `original` are **two names for one list**:

```text
original ──┐
           ├──►  [70, 80]   (the one and only list)
scores   ──┘
```

Here's a version that leaves the caller's list alone by building a **new** list (a Filter-style pattern that keeps everything):

````
def with_bonus(scores):
    result = []
    for s in scores:
        result.append(s + 5)
    return result

original = [70, 80]
curved = with_bonus(original)
print("curved:  ", curved)
print("original:", original)
````

### 🕵️ Which Ones Mutate?

For each line, predict: **does it change `nums`?** And **what does it return?**

```python
nums = [3, 1, 2]

a = sorted(nums)       # ?
b = nums.sort()        # ?
c = nums + [4]         # ?
d = nums.append(5)     # ?
```

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_

````
nums = [3, 1, 2]

a = sorted(nums)
print("sorted(nums) returned", a, "   nums is now", nums)

b = nums.sort()
print("nums.sort() returned", b, "   nums is now", nums)

c = nums + [4]
print("nums + [4] returned", c, "   nums is now", nums)

d = nums.append(5)
print("nums.append(5) returned", d, "   nums is now", nums)
````

**Rule of thumb:** methods that **change** a list in place (`.sort()`, `.append()`, `.remove()`) usually return `None`. Functions that **build something new** (`sorted()`, `+`) return the new thing and leave the original alone.

For your own functions, prefer **returning a new list** unless changing the original is the whole point. That way nobody gets surprised.
