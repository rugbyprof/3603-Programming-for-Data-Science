### 🔎 Anatomy of a Function

```text
 keyword   name    parameter
   │        │         │
   ▼        ▼         ▼
  def    square    ( x ) :       ← the header ends with a colon
      return x * x               ← the body is indented
      │
      └── return: send this value BACK to whoever called me
```

### 🧠 The Mental Model: A Machine

Think of a function as a box with an input slot and an output slot. You keep using it without knowing (or caring) what happens inside.

```text
              FUNCTION
           ┌────────────┐
   10 ───► │   square   │ ───► 100
           └────────────┘
          input           output
```

This is the model we'll use for the rest of the course. For **every** function you write or read, ask:

> **What goes in? What comes out?**

## 🔁 What Actually Happens During a Call

When Python reaches `square(10)`:

```text
1. Jump into the function.
2. Put the argument into the parameter:     x = 10
3. Run the body:                             x * x → 100
4. return sends 100 back.
5. The call  square(10)  is REPLACED by 100 where it was written.
```

Step 5 is the important one: **a function call becomes its return value.** That's why you can put a call anywhere you could put a number:

```text
square(3) + square(4)
    ↓           ↓
    9     +    16
          ↓
          25
```

## 🏷️ Parameter vs. Argument (5 minutes, then move on)

````
def square(number):
    return number * number

square(7)
````

```text
number  →  PARAMETER   the name in the definition (the empty slot)
7       →  ARGUMENT    the actual value you pass in when you call it
```

One way to remember it: the **p**arameter is the **p**laceholder, and the **a**rgument is the **a**ctual value.

That's all the terminology for now. We'll keep using these words correctly and you'll get used to them.

## 🧰 Back to the Real Problem

Now let's fix those three copies of the average code. **What changes between them?** Only the list. So the list becomes the parameter:

````
def average(numbers):
    total = 0
    for x in numbers:
        total += x
    return total / len(numbers)
````

It's the same Accumulate pattern you already know, just wrapped in a `def` with a name. Now all three averages take one line each:

````
scores = [87, 91, 74, 88, 95]
temps  = [31, 45, 28, 52, 60]
prices = [4.99, 12.50, 3.25]

print(average(scores))
print(average(temps))
print(average(prices))
````

```text
             ┌───────────┐
 [87, 91, …] │  average  │ ───► 87.0
   (a list)  └───────────┘     (a number)
```

**A container goes in and a single number comes out.** Most of the rest of this module is functions like this one: containers go in, and something useful comes out.

Now, what did we gain?

| Before | After |
|--------|-------|
| 3 copies of the loop | 1 copy |
| A bug has to be fixed in 3 places | Fix it once |
| 50 lists = 50 copies | 50 lists = 50 short lines |
| You have to *read* the loop to know what it does | The name `average` tells you |

## 🖨️ `print()` vs. `return` — Important!

These two look similar. They are **not** the same, and mixing them up is the #1 function bug in this course.

> **`print()` gives something to the human.**
> It shows it on the screen, and then it's gone.
>
> **`return` gives something back to the program.**
> The value can be stored, used in math, and passed on to other functions.

Here are two versions of `square`:

````
def square_print(x):
    print(x * x)

def square_return(x):
    return x * x
````

🔮 **Predict:** They look like they do the same thing. What does each block below do?

```python
a = square_return(10)
b = square_return(20)
print(a + b)
```

```python
a = square_print(10)
b = square_print(20)
print(a + b)
```

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_

````
a = square_return(10)
b = square_return(20)
print(a + b)
````

````
# ⚠️ This cell raises an error on purpose. Read the error message!
a = square_print(10)
b = square_print(20)
print(a + b)
````

What happened?

- `square_print(10)` **showed** `100` on the screen, but it didn't `return` anything.
- A function with no `return` gives back a special value called **`None`**, which means "nothing."
- So `a` is `None`, `b` is `None`, and `None + None` doesn't make sense. That's the error.

````
a = square_print(10)
print("a is:", a)
````

**Rule of thumb:** functions that *compute* something should `return` it. Do your `print()`ing at the end, outside the function, once you have the final answer.

## 🎛️ More Than One Parameter

A function can take several inputs. Arguments match up with parameters **by position**, so the first argument goes to the first parameter, and so on.

````
def count_above(numbers, threshold):
    count = 0
    for x in numbers:
        if x > threshold:
            count += 1
    return count

scores = [87, 91, 74, 88, 95]
print(count_above(scores, 90))
print(count_above(scores, 80))
````

```text
                     ┌──────────────┐
 [87, 91, 74, …] ──► │              │
                     │ count_above  │ ───► 2
            90   ──► │              │
                     └──────────────┘
```

## 💬 Talk About It: Why Functions *Really* Exist

"To avoid copy/paste" is the first reason. It isn't the only one:

1. **Reuse.** Write it once, use it everywhere.
2. **Naming.** `average(scores)` is easier to read than a 4-line loop. A good function name tells the reader what the code does.
3. **Fix once.** A bug in a function gets fixed in one place.
4. **Test in isolation.** You can check that `average` works by itself before you depend on it.
5. **Hiding details.** You use `len()` without knowing how it works. Other people (including you, next month) can use your functions the same way.
6. **Building blocks.** Small functions combine into bigger ones. That's the whole idea of Notebook 06.
