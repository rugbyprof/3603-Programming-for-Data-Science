- Using `%history` and `%save`

## 1. Modes in Jupyter

- **Edit mode** (cursor blinking in a cell, ready to type)
- **Command mode** (cell is selected, but you're not typing — shortcuts act on the whole cell)

Switch:
- Press `Enter` → Edit mode
- Press `Esc` → Command mode

## 2. Common Shortcuts

- `A` → Insert cell Above
- `B` → Insert cell Below
- `D D` → Delete a cell
- `M` → Change to Markdown cell
- `Y` → Change to Code cell
- `Shift + Enter` → Run cell and move to next
- `Ctrl + Enter` (`Cmd + Enter` on Mac) → Run cell, stay put
- `Alt + Enter` (`Option + Enter` on Mac) → Run cell and insert a new one below

## 3. Saving and Exporting

- `Ctrl + S` (`Cmd + S` on Mac) → Save the notebook
- Menu: *File → Save and Export Notebook As…* → export to `.py`, `.html`, or `.pdf`
- The workflow you're actually using in this course: save the notebook, then commit and push it with git/GitHub — see `_StartHere` if you need a refresher on that process.

```
# Save history to a file
%save my_session.py 1-5
```
