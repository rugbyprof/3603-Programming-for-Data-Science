## 1. Reading CSV Files

```
import pandas as pd

# Create a small CSV for demo
data = """Name,Age,Score
Alice,23,90
Bob,25,85
Charlie,22,95
"""
with open('students.csv', 'w') as f:
    f.write(data)

# Read CSV into a DataFrame
df = pd.read_csv('students.csv')
df
```
