## 3. JSON Files

````
import json

data = {'name': 'Graylian', 'type': 'mascot', 'skills': ['howl', 'parrot talk']}

# Write JSON
with open('data.json', 'w') as f:
    json.dump(data, f)

# Read JSON
with open('data.json', 'r') as f:
    loaded = json.load(f)
loaded
````

## 4. CSV Files

````
import csv

# Writing CSV
rows = [['Name', 'Score'], ['Alice', 90], ['Bob', 85]]
with open('scores.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open('scores.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

## 🔥 Challenge (Optional) — Paths That Work Anywhere
Typing `'example.txt'` only works because it happens to be in the current directory. The `pathlib` module (built into Python) gives you a more robust way to build paths that works the same on Windows, Mac, and Linux.
```python
from pathlib import Path

data_folder = Path('data')
file_path = data_folder / 'scores.csv'
print(file_path)
print(file_path.exists())
```
