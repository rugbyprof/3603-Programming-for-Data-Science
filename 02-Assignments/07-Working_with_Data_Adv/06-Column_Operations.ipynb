# 🧱 06 - Column Operations: Creating, Renaming, and Dropping

This notebook is self-contained — it builds its own sample data.

## ✅ Learning Goals
- Create new columns using calculations or functions
- Rename one or multiple columns
- Drop unwanted columns safely
- Apply a function across a column with `.apply()`

## 1. Setup and Sample Data

````
import pandas as pd

data = {
    'first_name': ['Alice', 'Bob', 'Charlie', 'David'],
    'last_name': ['Smith', 'Jones', 'Brown', 'Wilson'],
    'math_score': [85, 90, 78, 92],
    'science_score': [88, 85, 82, 95],
}

df = pd.DataFrame(data)
df
````

## 2. Creating New Columns

````
# Simple column arithmetic
df['average_score'] = (df['math_score'] + df['science_score']) / 2
df
````

````
# Combine two text columns
df['full_name'] = df['first_name'] + ' ' + df['last_name']
df
````

## 3. Renaming Columns

````
# Rename specific columns
df.rename(columns={'math_score': 'Math', 'science_score': 'Science'}, inplace=True)
df
````

````
# Rename ALL columns at once
df.columns = [col.upper() for col in df.columns]
df
````

## 4. Dropping Columns

````
df.drop(columns=['FULL_NAME'], inplace=True)
df
````

## 5. Applying a Function Across a Column
Remember `isinstance()`-based branching and defining your own functions from the Functions module? `.apply()` is where that pays off — you write a normal function, and Pandas runs it on every value in a column.

````
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    else:
        return 'C'

df['GRADE'] = df['AVERAGE_SCORE'].apply(grade)
df
````

So far `.apply()` has worked on a single column (a `Series`). It can also work across an *entire row* at once if you pass `axis=1` — each `row` your function receives is a mini `Series` you can index into by column name.
```python
def summarize(row):
    return f"{row['first_name']} scored {row['average_score']:.1f} on average"

df['summary'] = df.apply(summarize, axis=1)
```

---
### 📝 Summary
| Action              | Method                                 |
|---------------------|------------------------------------------|
| Create column       | `df['new'] = ...`                       |
| Rename column(s)    | `df.rename(columns={...})`              |
| Rename all columns  | `df.columns = [...]`                    |
| Drop column(s)      | `df.drop(columns=[...])`                |
| Apply function      | `df['col'].apply(func)`                 |
