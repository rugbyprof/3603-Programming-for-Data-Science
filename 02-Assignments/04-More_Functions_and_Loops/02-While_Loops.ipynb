# 📘 02 - While Loops

A `for` loop means **"do this for each item."** It works well when you already have the items.

But sometimes you don't know how many times you need to repeat something. You only know **when to stop**. That's what a `while` loop is for.

## ✅ Learning Goals
- Choose between `for` ("for each item") and `while` ("until something happens")
- Write a `while` loop with the three required parts: **start**, **condition**, **update**
- Recognize an infinite loop, and stop one
- Use `break` to stop early and `continue` to skip an item
- Wrap a `while` loop inside a function

## 😩 The Problem

You put **$1000** in a savings account that earns **5% per year**. How many years until it doubles?

You can't write `for year in ...` because you don't know how many years there will be. That's the question you're trying to answer. What you do know is **when to stop**: once the balance reaches $2000.

````
balance = 1000        # start
years = 0

while balance < 2000:        # condition: keep going WHILE this is True
    balance = balance * 1.05 # update
    years += 1

print(years, "years →", round(balance, 2))
````

### 🔎 Anatomy

Every `while` loop needs **three** parts. If one is missing, the loop is broken.

```text
balance = 1000               ← 1. START:     set up the variable the condition checks
while balance < 2000:        ← 2. CONDITION: checked BEFORE every pass
    balance = balance * 1.05 ← 3. UPDATE:    must eventually make the condition False
    years += 1
```

## 🚫 Infinite Loops

If the **update** is missing (or never makes the condition `False`), the loop runs forever:

```python
n = 1
while n < 20:
    print(n)        # nothing ever changes n
```

This happens to **everyone**. If a cell keeps running and the notebook seems stuck, click the **Interrupt / Stop** button next to the cell (or use *Kernel → Interrupt*), fix the update, and run it again.

## ⚖️ `for` or `while`?

| Use `for` when... | Use `while` when... |
|--------------------|----------------------|
| you have a collection of items | you're waiting for something to happen |
| "for each score..." | "until the balance doubles..." |
| the loop ends when the items run out | *you* have to make the loop end |

**When in doubt, use `for`.** It can't loop forever.

## 🛑 `break` — Stop Early

**Problem:** find the **first** failing score (below 60), and don't check any more after that.

````
scores = [87, 91, 55, 88, 42]

for s in scores:
    if s < 60:
        print("first failing score:", s)
        break                # leave the loop immediately
    print("checked", s)
````

Does this look familiar? It's the same idea as `mystery4` in Module 03b, where `return` left the function as soon as it found an answer. `break` leaves the **loop**. `return` leaves the whole **function**.

## ⏭️ `continue` — Skip This One

`continue` skips the rest of the body **for this item only** and moves on to the next one. It's handy for ignoring bad data:

````
readings = [72, -999, 75, 71, -999, 80]   # -999 means "sensor error"

total = 0
count = 0
for r in readings:
    if r == -999:
        continue             # skip bad readings
    total += r
    count += 1

print("average of good readings:", total / count)
````

## 🚦 Sentinel Values — "Stop When You See This"

A **sentinel** is a special value that means "the data ends here." For example, a data logger might write readings until it hits `-1`:

````
readings = [72, 75, 71, -1, 80, 90]   # everything after -1 is garbage

good = []
i = 0
while readings[i] != -1:
    good.append(readings[i])
    i += 1

print(good)
````

🔮 What would go wrong if `readings` **didn't** contain a `-1`? Predict, then think about how you'd protect against it. (Hint: the condition could check two things with `and`.)

### 🔀 Cross-Language Note: No `do`/`while` in Python

C++ has `do { ... } while (cond);`, which always runs the body at least once. Python doesn't have it. The usual substitute is `while True:` with a `break`:

```python
while True:
    n = int(input("Enter a positive number: "))
    if n > 0:
        break
```
