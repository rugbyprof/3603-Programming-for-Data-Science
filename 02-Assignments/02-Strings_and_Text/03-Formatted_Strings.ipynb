# 📘 03 - Formatted Strings (f-strings)

A closer look at f-strings — the modern, preferred way to build strings out of variables in Python.

## ✅ Learning Goals
- Embed multiple expressions in a single f-string
- Control number and text formatting with format specs
- Call methods and functions directly inside `{}`
- Recognize older formatting styles you might see in existing code

## 🧵 The Basics, Revisited
You've already used f-strings a little in Scalar Types. An f-string is a normal string with an `f` right before the opening quote — anything inside `{}` gets evaluated and inserted.

````python
name = 'Ada'
age = 28
print(f'{name} is {age} years old')
````

## 🔢 Multiple Expressions in One String
You're not limited to one `{}`, and you're not limited to plain variables — any expression works inside the braces.

````python
price = 19.99
qty = 3
print(f'{qty} items at ${price} each = ${price * qty}')
````

## 🎯 Format Specs
After a `:` inside the `{}`, you can control exactly how a value is displayed — decimal places, thousands separators, and alignment.

````python
pi = 3.14159265
total = 1234567.891
print(f'{pi:.2f}')          # 2 decimal places
print(f'{total:,.2f}')      # thousands separator + 2 decimals
print(f'{"hi":>10}|')      # right-align in a 10-char field
print(f'{"hi":<10}|')      # left-align in a 10-char field
print(f'{"hi":^10}|')      # center in a 10-char field
````

## 🧠 Calling Methods and Functions Inside `{}`

````python
name = 'ada'
print(f'{name.upper()} has {len(name)} letters')
````

## 📜 A Quick History (So You Recognize Older Code)
Before f-strings (Python 3.6+), people used `.format()` or `%`-style formatting. You'll still run into both in older code and online examples, so it's worth recognizing them even though f-strings are what you should write going forward.

````python
name = 'Ada'
age = 28
print('{} is {} years old'.format(name, age))   # .format() style
print('%s is %d years old' % (name, age))       # %-style (oldest)
print(f'{name} is {age} years old')              # f-string (preferred)
````
