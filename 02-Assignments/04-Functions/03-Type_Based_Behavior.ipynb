# 📘 03 - Type-Based Behavior

Python is dynamically typed — a variable can hold any type, and a function doesn't know in advance what type it'll be handed. Sometimes that means writing functions that check what they got and behave differently.

## ✅ Learning Goals
- Use `type()` and `isinstance()` to check a value's type inside a function
- Understand why this is Python's stand-in for "function overloading"
- Recognize duck typing as an alternative to explicit type checks

## 🤔 The Problem
In a language like C++, you can define two functions with the *same name* but *different* parameter types (overloading), and the compiler picks the right one at compile time. Python doesn't work that way — there's only ever one `def` with a given name in a scope. If you want a function to behave differently depending on what type it receives, you have to check the type yourself, at runtime.

```
def double(x):
    return x * 2

print(double(5))         # 10
print(double('5'))       # '55' — technically works, just very differently
print(double([1, 2]))    # [1, 2, 1, 2] — also "works", very differently again!
```

## 🔍 Checking Type with `type()` and `isinstance()`
`isinstance()` is usually the better choice for checks like this — it also recognizes subclasses, which `type(x) == int` does not.

```
def describe_type(x):
    if isinstance(x, int):
        print(f'{x} is an integer')
    elif isinstance(x, str):
        print(f'{x!r} is a string')
    elif isinstance(x, list):
        print(f'{x} is a list with {len(x)} items')
    else:
        print(f'{x} is a {type(x).__name__}')

describe_type(5)
describe_type('hello')
describe_type([1, 2, 3])
describe_type(3.14)
```

## 🎭 Python's Stand-In for Overloading
Since you can't have two `def add(...)` with different parameter types, you branch on the type *inside* the one function you're allowed to have.

```
def add(a, b):
    if isinstance(a, str) or isinstance(b, str):
        return f'{a}{b}'   # string-ish behavior: concatenate
    return a + b            # numeric behavior: add

print(add(2, 3))                 # 5
print(add('2', '3'))             # '23'
print(add('Data ', 'Science'))   # 'Data Science'
```

### Same Idea as `dict.get()`
This is the same underlying pattern you've been using all along: check what you actually have, and choose behavior based on it.

- `dictionary.get(key, default)` asks: *does this key exist?* → pick a value accordingly.
- A type check asks: *is this the right type?* → pick a behavior accordingly.

```
student = {}
major = student.get('major', 'Undeclared')   # "does the key exist?" check
print(major)

def label(value):
    return 'number' if isinstance(value, (int, float)) else 'not a number'   # "right type?" check

print(label(42))
print(label('42'))
```

## 🦆 Duck Typing — The Pythonic Alternative
"If it walks like a duck and quacks like a duck..." Often, instead of checking a value's type explicitly, Python code just *tries* to use it the way it expects, and handles the failure if that doesn't work. This is common enough to have a name: **duck typing**.

```
def get_length(thing):
    try:
        return len(thing)
    except TypeError:
        return 'no length'

print(get_length('hello'))     # 5
print(get_length([1, 2, 3]))   # 3
print(get_length(42))          # 'no length'
```

This isn't your only tool for this, and you won't need `try`/`except` mastered today — it's just worth recognizing. Later in the course, this exact pattern shows up constantly in pandas, where a function might accept a single column or a whole DataFrame and behave sensibly either way.
