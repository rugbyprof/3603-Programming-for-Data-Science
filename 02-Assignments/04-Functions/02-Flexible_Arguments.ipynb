# 📘 02 - Flexible Arguments

Not every function knows in advance how many arguments it needs. Python gives you tools to handle that.

## ✅ Learning Goals
- Tell the difference between positional and keyword arguments
- Write functions that accept any number of positional arguments with `*args`
- Write functions that accept any number of keyword arguments with `**kwargs`
- Mix regular parameters, `*args`, and `**kwargs` in one function

## 📍 Positional vs. Keyword Arguments

````
def introduce(name, age):
    print(f'{name} is {age} years old')

introduce('Ada', 28)              # positional — order matters
introduce(age=28, name='Ada')     # keyword — order doesn't matter
````

## 📦 `*args` — Any Number of Positional Arguments
You saw `*args` in the Tuples module as a leftover collector (`first, *rest = ...`). Here's the other half: you can define your *own* function that accepts any number of arguments this way.

````
def total(*args):
    print(type(args), args)
    return sum(args)

print(total(1, 2, 3))
print(total(1, 2, 3, 4, 5))
````

## 🗃️ `**kwargs` — Any Number of Keyword Arguments
`**kwargs` is the same idea, but for *named* arguments — Python collects them into a dictionary.

````
def describe_pet(**kwargs):
    print(type(kwargs), kwargs)
    for key, value in kwargs.items():
        print(f'{key}: {value}')

describe_pet(name='Rex', species='Dog', age=3)
````

## 🧩 Mixing Them Together
Python requires a specific order: regular parameters first, then `*args`, then `**kwargs`.

````
def build_profile(name, *hobbies, **details):
    print('Name:', name)
    print('Hobbies:', hobbies)
    print('Other details:', details)

build_profile('Ada', 'math', 'writing', occupation='Mathematician', era='1800s')
````

## 🔥 Challenge (Optional) — Unpacking *Into* a Function Call
Just as `*args`/`**kwargs` collect arguments **into** a function, you can use `*`/`**` to unpack a list or dict **out into** a function call.
```python
def add3(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add3(*nums))     # unpacks the list into 3 positional arguments

info = {'a': 1, 'b': 2, 'c': 3}
print(add3(**info))    # unpacks the dict into keyword arguments
```
Try it: build a list or dict of your own and use `*`/`**` to call one of your earlier functions with it.

````
def add3(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add3(*nums))

info = {'a': 1, 'b': 2, 'c': 3}
print(add3(**info))
````
