# 📘 03 - Python Dictionaries

Dictionaries store key-value pairs for fast lookups.

## ✅ Learning Goals
- Create and access dictionary items
- Use dictionary methods like `.get()` and `.items()`

````
# 📚 Create a Dictionary
student = {'name': 'Alex', 'age': 21, 'major': 'CS'}
print(student)
````

````
# 🔍 Access Values
print(student['name'])
print(student.get('gpa', 'N/A'))
````

````
# 🔁 Loop Over Dictionary
for key, value in student.items():
    print(key, '->', value)
````

## 📋 A List of Dictionaries (rows in a table)

This is the single most common data shape you'll meet in data science. Each **dictionary is one row** (one record), and the **list holds all the rows**. Every dictionary uses the same keys, which act like column headers.

````
students = [
    {'name': 'Alex', 'age': 21, 'major': 'CS'},
    {'name': 'Ana',  'age': 20, 'major': 'Math'},
    {'name': 'Ben',  'age': 22, 'major': 'History'},
    {'name': 'Cy',   'age': 19, 'major': 'Biology'},
]

# Loop over the rows
for row in students:
    print(f"{row['name']:6} | age {row['age']} | {row['major']}")

print()
print('Number of students:', len(students))
print('Second student:', students[1])            # index into the list
print('Second student\'s major:', students[1]['major'])   # then key into the dict

# Add a new row
students.append({'name': 'Dana', 'age': 23, 'major': 'CS'})

# Filter: everyone majoring in CS
cs_students = [row['name'] for row in students if row['major'] == 'CS']
print('CS majors:', cs_students)
````

## 🔑 Using the Row Number as a Key

Sometimes you want to look a record up by its **row number** instead of scanning the whole list. Wrapping the list in a dictionary keyed by row number gives you an instant lookup — and because the value is still a dictionary, you can reach in and update any field (like changing your major).

````
roster = [
    {'name': 'Alex', 'age': 21, 'major': 'CS'},
    {'name': 'Ana',  'age': 20, 'major': 'Math'},
    {'name': 'Ben',  'age': 22, 'major': 'History'},
    {'name': 'Cy',   'age': 19, 'major': 'Biology'},
]

# Build a dict:  row number -> record
by_row = {i: row for i, row in enumerate(roster)}
print('Row 2:', by_row[2])

# Find and update a value by row number: Ben (row 2) switches to CS
by_row[2]['major'] = 'CS'
print('Row 2 after change:', by_row[2])

# by_row[2] is the SAME dictionary object that lives in roster,
# so the original list sees the change too
print('roster[2]:', roster[2])

# A helper that updates any field for any row
def update_field(table, row_num, key, value):
    table[row_num][key] = value

update_field(by_row, 0, 'major', 'Data Science')
print('Row 0 after update:', by_row[0])
````

## 🌐 JSON and GeoJSON

**JSON** (JavaScript Object Notation) is the most common format for sharing data on the web. The syntax is almost identical to Python:

| JSON | Python |
|------|--------|
| object `{ }` | `dict` |
| array `[ ]` | `list` |
| string `"..."` | `str` (double quotes only) |
| number | `int` / `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

So a JSON file is really just **nested lists and dictionaries**. Python's built-in `json` module converts between the two: `json.loads()` (text → Python) and `json.dumps()` (Python → text).

**GeoJSON** is JSON with an agreed-upon structure for geographic data: a `FeatureCollection` holds a list of `Feature` objects, each with a `geometry` (a `Point`, `LineString`, `Polygon`, ... plus its coordinates) and a free-form `properties` dictionary. Because it's just JSON, you read it with the same list/dict indexing.

````
import json

# A JSON string (note: double quotes, true/null instead of True/None)
raw = '''
{
  "course": "Programming for Data Science",
  "credits": 3,
  "online": true,
  "instructor": null,
  "students": [
    {"name": "Alex", "major": "CS"},
    {"name": "Ana",  "major": "Math"}
  ]
}
'''

data = json.loads(raw)          # JSON text -> Python dict/list
print(type(data))              # <class 'dict'>
print(data['course'])
print(data['students'][1]['name'])   # list index, then dict key
print('online?', data['online'])     # -> True (Python bool)

# Go the other way: Python -> JSON text
print(json.dumps({'name': 'Cy', 'grades': [88, 91]}, indent=2))
````

````
# A tiny GeoJSON FeatureCollection: two campus buildings as points
geo = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "geometry": {"type": "Point", "coordinates": [-98.529, 33.878]},
            "properties": {"name": "Bolin Hall", "floors": 4},
        },
        {
            "type": "Feature",
            "geometry": {"type": "Point", "coordinates": [-98.531, 33.876]},
            "properties": {"name": "Moffett Library", "floors": 3},
        },
    ],
}

# It's just nested lists and dicts, so normal indexing walks it
for feature in geo["features"]:
    name = feature["properties"]["name"]
    lon, lat = feature["geometry"]["coordinates"]   # GeoJSON is [longitude, latitude]
    print(f"{name:16} at lon={lon}, lat={lat}")
````
