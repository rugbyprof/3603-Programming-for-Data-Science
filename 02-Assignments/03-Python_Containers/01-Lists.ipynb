## ➕ Combining Lists: `+` vs `append()` vs `extend()`

- `list_a + list_b` builds a **new** list, leaving the originals unchanged
- `append(x)` adds `x` as a **single element** (a nested list if `x` is a list)
- `extend(iterable)` adds **each item** of the iterable to the end, modifying the list in place

````
a = [1, 2, 3]
b = [4, 5, 6]

# 1) Adding with + makes a brand new list
combined = a + b
print('a + b        ->', combined)
print('a unchanged  ->', a)

# 2) append() adds ONE element (here, the whole list b becomes a single item)
a_append = [1, 2, 3]
a_append.append(b)
print('append(b)    ->', a_append)   # [1, 2, 3, [4, 5, 6]]

# 3) extend() adds each item of b individually
a_extend = [1, 2, 3]
a_extend.extend(b)
print('extend(b)    ->', a_extend)   # [1, 2, 3, 4, 5, 6]

# extend() also works with any iterable, like a string
letters = ['a', 'b']
letters.extend('cd')
print('extend("cd") ->', letters)    # ['a', 'b', 'c', 'd']
````

## 🔲 Multidimensional Lists

A list can hold other lists, which gives you a **grid** (rows and columns). Use two indexes to reach a single value: `grid[row][col]`.

````
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print('Whole grid:', grid)
print('Row 0:     ', grid[0])
print('Row 1, col 2:', grid[1][2])   # -> 6

# Update a single cell
grid[0][0] = 99
print('After change:', grid)

# Walk every row and column
for row in grid:
    for value in row:
        print(value, end=' ')
    print()   # newline after each row
````

## 🗑️ Deleting From a List

Several ways to remove items:

- `remove(value)` — deletes the **first** matching value
- `pop(index)` — removes the item at `index` and **returns it** (defaults to the last item)
- `del mylist[index]` — deletes by index (or a slice); no value returned
- `clear()` — empties the list

````
items = ['a', 'b', 'c', 'd', 'e', 'f']

items.remove('c')          # delete by value
print("after remove('c'):", items)

last = items.pop()         # remove & return the last item
print('popped last:', last, '->', items)

second = items.pop(1)      # remove & return index 1
print('popped index 1:', second, '->', items)

del items[0]               # delete by index
print('after del items[0]:', items)

items.clear()              # remove everything
print('after clear():', items)
````

## 🔢 Iterating With an Index: `for i in range(len(mylist))`

Looping directly (`for item in mylist`) is the usual choice, but sometimes you need the **position** of each item — for example to print numbered lines or to modify the list as you go. `range(len(mylist))` gives you `0, 1, 2, ...` up to the last valid index.

````
movies = ['Alien', 'Arrival', 'Dune', 'Gattaca']

# Use the index to print a numbered list
for i in range(len(movies)):
    print(i + 1, '-', movies[i])

# The index lets you change items in place
prices = [10, 20, 30, 40]
for i in range(len(prices)):
    prices[i] = prices[i] * 1.1     # add 10%
print('with tax:', prices)

# Same idea with enumerate(), which hands you index AND value together
for i, movie in enumerate(movies):
    print(f'{i}: {movie}')
````
