## 3. Box Plots for Outliers
Box plots are great for spotting outliers.

````
sns.boxplot(data=penguins, x='species', y='bill_length_mm',
            hue='species', palette='Set2', legend=False)
plt.title('Bill Length by Species')
plt.show()
````

## 4. Correlation Heatmap
Correlation heatmaps let us see every pairwise relationship between numeric variables at once.

````
corr = penguins.dropna().corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()
````
