# 🐧 04 - Seaborn Basics

Seaborn is a **high-level visualization library** built on top of Matplotlib. It makes it easier to create attractive, statistically-informed plots with less code.

This notebook covers:
- Why Seaborn?
- Common plot types (scatter, bar, histogram, box)
- Styling and color palettes

## 1. Importing Seaborn and Loading Data
This module's folder includes `penguins.csv` locally, so this works offline — no download step, and no dependency on Seaborn's online example-dataset repository.

````
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

penguins = pd.read_csv('penguins.csv')
penguins.head()
````

👉 Seaborn *can* also load its own built-in example datasets with `sns.load_dataset('penguins')`, which is where this file originally came from — but that downloads over the network every time, so we're using a local copy instead.

## 2. Scatter Plot
Seaborn makes scatterplots simple — and `hue` colors points by category automatically.

````
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
plt.title('Penguins: Flipper vs Body Mass')
plt.show()
````

## 3. Bar Plot
Bar plots are also straightforward. `errorbar='sd'` shows a standard-deviation error bar on top of each bar (the current Seaborn spelling — older tutorials online may show `ci=`, which was renamed).

````
sns.barplot(data=penguins, x='species', y='body_mass_g', errorbar='sd')
plt.title('Average Body Mass by Species')
plt.show()
````

## 4. Histogram and KDE
Seaborn can combine a histogram with a kernel density estimate (KDE) — a smoothed curve showing the same shape.

````
sns.histplot(data=penguins, x='body_mass_g', kde=True, hue='species')
plt.title('Distribution of Body Mass')
plt.show()
````

## 5. Box Plot
Box plots show the median, quartiles, and outliers at a glance. Passing the same column to both `x` and `hue` (with `legend=False`) is the current recommended way to color a boxplot by category — this is the modern Seaborn spelling of what used to just be `palette=` on its own.

````
sns.boxplot(data=penguins, x='species', y='flipper_length_mm',
            hue='species', palette='Set2', legend=False)
plt.title('Flipper Length by Species')
plt.show()
````

## 6. Styling and Palettes
Seaborn makes it easy to set a theme and color palette for every plot that follows.

````
sns.set_style('whitegrid')
sns.set_palette('pastel')

sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species')
plt.title('Styled Scatterplot')
plt.show()
````
