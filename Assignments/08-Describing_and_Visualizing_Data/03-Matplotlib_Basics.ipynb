# 🎨 03 - Matplotlib Basics

Matplotlib is the **foundational plotting library** in Python. It's powerful (but sometimes verbose) — think of it as the "engine" that other libraries (like Seaborn) build on top of.

This notebook covers:
- Anatomy of a plot (`figure`, `axes`)
- Line, bar, scatter, and histogram basics
- Labels, titles, legends, and gridlines
- Customizing colors and styles

## 1. Importing Matplotlib
The most common convention is:

````
import matplotlib.pyplot as plt
import numpy as np

# Simple data
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.show()
````

👉 The `.plot()` function is the workhorse for line charts.

## 2. Anatomy of a Plot
- **Figure**: the whole canvas
- **Axes**: an individual plot within the figure

You can be explicit about creating them:

````
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title('A Sine Wave')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
plt.show()
````

👉 Using `ax.set_*` is cleaner when making multiple plots, since each `ax` keeps its own settings separate.

## 3. Basic Plot Types

### Line Plot

````
plt.plot(x, y, label='sin(x)', color='blue')
plt.plot(x, np.cos(x), label='cos(x)', color='red')
plt.title('Line Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
````

### Bar Plot

````
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 2]

plt.bar(categories, values, color='purple')
plt.title('Bar Plot Example')
plt.show()
````

### Scatter Plot

````
np.random.seed(42)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)

plt.scatter(x_scatter, y_scatter, color='green', marker='o')
plt.title('Scatter Plot Example')
plt.show()
````

### Histogram

````
hist_data = np.random.randn(1000)
plt.hist(hist_data, bins=30, color='orange', edgecolor='black')
plt.title('Histogram Example')
plt.show()
````

## 4. Customizing Styles
Matplotlib ships with several built-in styles you can switch to instantly.

````
plt.style.available[:10]  # show some available styles
````

````
plt.style.use('ggplot')
plt.plot(np.cumsum(np.random.randn(100)))
plt.title('Styled Plot (ggplot)')
plt.show()

plt.style.use('default')  # reset to default
````
