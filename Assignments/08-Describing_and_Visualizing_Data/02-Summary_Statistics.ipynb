# 📈 02 - Summary Statistics

Summary statistics let us describe a dataset with just a few numbers. Instead of staring at thousands of rows, we can quickly understand **center, spread, and shape**.

This notebook covers:
- Measures of central tendency (mean, median, mode)
- Measures of spread (range, variance, standard deviation, IQR)
- Skewness and kurtosis
- Quick summaries with Pandas

## 1. Central Tendency
These describe the "middle" of the data:
- **Mean** — the average
- **Median** — the middle value when sorted
- **Mode** — the most frequent value

````
import numpy as np
import pandas as pd
from scipy import stats

data = [5, 7, 8, 5, 10, 12, 7, 7, 6, 9]

print('Mean:', np.mean(data))
print('Median:', np.median(data))
print('Mode:', stats.mode(data, keepdims=True)[0][0])
````

👉 **Question:** Which is more robust to outliers, the mean or the median? Write your answer, then verify it in the cell below by adding an extreme value to `data` and re-running the mean/median calculation.

````
data_with_outlier = data + [100]
print('Mean:', np.mean(data_with_outlier))
print('Median:', np.median(data_with_outlier))
````

## 2. Spread
Spread tells us how variable the data are.
- **Range**: max − min
- **Variance**: average squared deviation from the mean
- **Standard Deviation**: square root of variance (back in the original units)
- **Interquartile Range (IQR)**: the middle 50% (Q3 − Q1)

````
print('Range:', np.max(data) - np.min(data))
print('Variance:', np.var(data, ddof=1))
print('Standard Deviation:', np.std(data, ddof=1))
print('IQR:', stats.iqr(data))
````

👉 **Exercise:** Using the `data_with_outlier` list from above, compare the standard deviation with and without the outlier. How much did one extra value change it?

````
print('Std Dev (no outlier):', np.std(data, ddof=1))
print('Std Dev (with outlier):', np.std(data_with_outlier, ddof=1))
````

### 🧠 Quick Check
`np.var()` and `np.std()` above both use `ddof=1`. Look up what `ddof` stands for — why might you want `ddof=1` for a *sample* instead of the default `ddof=0`, which assumes you have the *entire population*?

## 3. Shape: Skewness and Kurtosis
- **Skewness**: measures asymmetry (which tail is longer)
- **Kurtosis**: measures "peakedness" or how heavy the tails are

Most real-life datasets are *not* perfectly normal, so these help describe how they differ.

````
print('Skewness:', stats.skew(data))
print('Kurtosis:', stats.kurtosis(data))
````

👉 **Note:** High kurtosis means more extreme outliers are likely; low kurtosis means flatter, more "boring" data.

## 4. Quick Summaries with Pandas
Instead of writing out every function above by hand, Pandas does it all at once with `.describe()`.

````
df = pd.DataFrame({'Values': data})
df.describe()
````
