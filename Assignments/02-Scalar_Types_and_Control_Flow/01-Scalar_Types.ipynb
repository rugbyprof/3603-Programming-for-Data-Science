# 📘 01 - Scalar Types

Review and practice working with Python's basic scalar data types.

## ✅ Learning Goals
- Recognize scalar types (`int`, `float`, `str`, `bool`)
- Use the `type()` function to identify types, and convert between types with casting
- Perform arithmetic operations and understand what each operator does
- Use variables, assignment, and comparison operators

## 🔧 Define Some Variables

````
num = 42
pi = 3.14159
name = 'Python'
is_cool = True
````

## 🔍 Check Their Types

````
print(type(num))
print(type(pi))
print(type(name))
print(type(is_cool))
````

## ➕ Arithmetic Operators
Python supports the usual math operators, plus a couple that might be new to you: `//` (floor division, drops the remainder) and `**` (exponent).

````
x = 10
y = 3
print('Add:', x + y)
print('Subtract:', x - y)
print('Multiply:', x * y)
print('Divide:', x / y)
print('Floor Divide:', x // y)
print('Remainder (Modulo):', x % y)
print('Exponent:', x ** 2)
````

## 🔄 Type Casting
Sometimes you need to convert a value from one type to another — for example, `input()` always returns a `str`, even if the user types a number.

````
age_text = '21'
age_number = int(age_text)
print(age_number + 1)
print(type(age_number))
````
