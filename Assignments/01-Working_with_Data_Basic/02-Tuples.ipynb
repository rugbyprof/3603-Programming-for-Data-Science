## 🧵 Why Tuples Matter (Beyond Just 'Immutable Lists')
Immutability isn't just a restriction — it's what makes tuples **hashable**, which unlocks things lists can't do. That's a big part of why you'll see tuples all over real data science and CS libraries.

### Tuples as Dictionary Keys
Because tuples can't change, Python can use them as dictionary keys or put them in a `set`. Lists can't do this.

````
locations = {}
locations[(29.4241, -98.4936)] = 'San Antonio'
locations[(32.7767, -96.7970)] = 'Dallas'
print(locations[(29.4241, -98.4936)])

# Uncomment to see why lists can't do this:
# locations[[29.4241, -98.4936]] = 'San Antonio'
````

### Multiple Return Values
Ever wonder how a function "returns two things"? It's actually returning one tuple, which you then unpack.

````
def min_max(nums):
    return min(nums), max(nums)

lo, hi = min_max([4, 8, 1, 9, 3])
print(lo, hi)
````

### `zip()` Produces Tuples
`zip()` pairs up items from two sequences — and each pair comes back as a tuple.

````
names = ['Ana', 'Ben', 'Cy']
scores = [90, 85, 77]
for pair in zip(names, scores):
    print(pair)
````

### `*args` Collects into a Tuple
When a function accepts `*args`, Python gathers the extra positional arguments into a tuple.

````
def total(*args):
    print(type(args), args)
    return sum(args)

print(total(1, 2, 3, 4))
````

## 🔥 Challenge (Optional)
Tuple unpacking has a neat trick: swapping two variables without a temporary variable.

```python
a, b = 1, 2
a, b = b, a
print(a, b)  # 2 1
```

For a step further, Python's `collections.namedtuple` lets you give each position in a tuple a name — handy when a plain tuple like `(5, 10)` doesn't explain itself.

```python
from collections import namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(5, 10)
print(p.x, p.y)
```
