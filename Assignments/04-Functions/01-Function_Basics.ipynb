# 📘 01 - Function Basics

Functions let you package up a piece of logic once and reuse it, instead of copying and pasting the same code every time you need it.

## ✅ Learning Goals
- Define a function with `def`, parameters, and `return`
- Understand the difference between `return` and `print`
- Return multiple values from one function (as a tuple)
- Use default parameter values
- Understand local vs. global variable scope

## 🔧 Defining a Function

````
def greet(name):
    return f'Hello, {name}!'

message = greet('Ada')
print(message)
````

### `return` vs. `print`
A function that only `print()`s something shows it on screen but hands nothing back to whoever called it. A function that `return`s something hands back a real value you can store, pass along, or use in more calculations.

````
def add_and_print(a, b):
    print(a + b)   # shows on screen, but gives the caller nothing to use

def add_and_return(a, b):
    return a + b   # this can actually be used afterward

result1 = add_and_print(2, 3)
result2 = add_and_return(2, 3)
print('result1 is:', result1)
print('result2 is:', result2)
````

## 🔁 Returning Multiple Values
Remember from Tuples: a function that looks like it "returns two things" is actually returning one tuple. You can unpack it right away.

````
def min_max(nums):
    return min(nums), max(nums)

lo, hi = min_max([4, 8, 1, 9, 3])
print(lo, hi)
````

## 🎛️ Default Parameter Values
You've already used this idea with `dictionary.get(key, default)` — if a value isn't given, a fallback is used instead. You can build the same pattern into your own functions.

````
def greet(name, greeting='Hello'):
    return f'{greeting}, {name}!'

print(greet('Ada'))
print(greet('Ada', 'Hi'))
print(greet('Ada', greeting='Welcome'))
````

### The Same Idea, Two Places
Compare these two lines — they're solving the exact same problem: *use this value, unless something more specific is available.*

````
student = {'name': 'Ada'}
major = student.get('major', 'Undeclared')   # dict default
print(major)

def describe(name, major='Undeclared'):        # function default
    return f'{name} — {major}'

print(describe('Ada'))
````

## 🌐 Local vs. Global Scope
A variable created inside a function only exists inside that function — it's *local*. It doesn't leak out, and assigning to it doesn't touch a variable of the same name outside.

````
x = 10

def change_x():
    x = 99   # this creates a NEW local variable, not the outer one
    print('Inside function, x =', x)

change_x()
print('Outside function, x =', x)
````
