# 📂 03 - Loading and Exploring Data

Welcome to the real world — where data lives in messy CSVs and your job is to make sense of it.

In this notebook, you'll:
- Load a CSV file with `pd.read_csv()`
- Use `.head()`, `.tail()`, `.info()`, and `.describe()` to explore it
- Identify potential issues (missing values, bad types, oddball rows)

## ✅ Learning Goals
- Load a local CSV into a DataFrame
- Preview and summarize a DataFrame's structure
- Count and locate missing values
- Rename a column for clarity

````
import pandas as pd
````

## 📥 Load a CSV
This module's folder includes `titanic.csv` — the classic Titanic passenger dataset — so everything here works offline, with no download step.

````
df = pd.read_csv('titanic.csv')
df.head()
````

## 📊 Peek at the Data

````
print('First 5 rows:')
print(df.head())

print('\nLast 5 rows:')
print(df.tail())
````

## 🧠 Understand the Structure

````
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
````

## 🧼 Data Types and Missing Values

````
print('Info:')
df.info()

print('\nMissing values per column:')
print(df.isnull().sum())
````

## 📈 Summary Statistics

````
df.describe(include='all')
````

## ✏️ Renaming Columns

````
# Rename 'Sex' to 'Gender' for clarity
df.rename(columns={'Sex': 'Gender'}, inplace=True)
df.head(2)
````
