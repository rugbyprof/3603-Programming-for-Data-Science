## 🎯 Sets — Unordered, Unique Values Only
A `set` is like a list that automatically throws away duplicates and doesn't care about order. Create one with `{}` (like a dict, but without `:` pairs) or `set(...)`.

```
courses = {'CS101', 'CS102', 'CS103', 'CS101'}  # duplicate 'CS101' is silently ignored
print(courses)
print('Is CS102 in the set?', 'CS102' in courses)
```

### Adding and Removing

```
courses.add('CS104')
courses.discard('CS101')   # .discard() doesn't error if the item isn't there
print(courses)
```

### Why Not Just Use a List?
Two big reasons: **deduplication** (turning a list with repeats into unique values is one line) and **fast membership checks** (`in` on a large set is dramatically faster than `in` on a large list, since sets are hash-based — the same mechanism that makes a dict lookup fast, from the Dictionaries module).

```
raw_signups = ['ana', 'ben', 'ana', 'cy', 'ben', 'ana']
unique_signups = set(raw_signups)
print('Raw:', raw_signups)
print('Unique:', unique_signups)
print('How many unique people signed up?', len(unique_signups))
```

## 🔁 Set Operations
Sets support the math operations you'd expect from Venn diagrams.

```
morning_class = {'Ana', 'Ben', 'Cy', 'Deb'}
afternoon_class = {'Cy', 'Deb', 'Eli', 'Fay'}

print('In either class (union):', morning_class | afternoon_class)
print('In both classes (intersection):', morning_class & afternoon_class)
print('Only in morning (difference):', morning_class - afternoon_class)
print('In exactly one class (symmetric difference):', morning_class ^ afternoon_class)
```

## 🔥 Challenge (Optional) — Sets vs. Lists for Membership Speed
Use `%timeit` (from the Foundations module) to compare how long `x in big_list` takes versus `x in big_set`, where both contain the same 100,000 numbers and `x` is a value near the end (the slowest case for a list).

```
big_list = list(range(100_000))
big_set = set(range(100_000))

%timeit 99_999 in big_list
%timeit 99_999 in big_set
```
