## 3. The `help()` Function
`help()` is the plain built-in Python equivalent of `?` — it works the same way outside of Jupyter too, e.g. in a plain `.py` script run from the terminal.

````
help(len)
````

````
import math
help(math.sqrt)
````

## 4. Tab Completion

In a Jupyter cell, type an object name followed by a dot and press `Tab` to see available methods.

Example:
```python
mylist = []
mylist.<TAB>
```
will show you all methods like `append`, `clear`, `copy`, etc.
