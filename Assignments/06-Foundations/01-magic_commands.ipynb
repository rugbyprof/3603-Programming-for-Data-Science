# 🧙 01 - Magic Commands

This notebook introduces some of the most useful **IPython magic commands** in Jupyter. Run each cell and see what happens!

## ✅ Learning Goals
- Use `%pwd`, `%ls`, `%cd` to navigate the filesystem from inside a notebook
- Inspect variables in memory with `%who` / `%whos`
- Write and run a `.py` file from inside a notebook
- Capture output and render HTML/Markdown from a code cell
- Discover more magics with `%lsmagic` and `?`

## 4. Writing and Running Files
`%%writefile` (a **cell** magic — note the double `%%`) saves the entire cell's contents to a file instead of running it as Python.

```
%%writefile hello.py
print('Hello from a file!')
```

```
%run hello.py
```

## 5. Output and Formatting

```
%%capture cap
print('This will not display right away')
```

```
cap.show()
```

```
%%html
<h1 style="color: maroon">Magics are Magical ✨</h1>
```

## 6. Discover More
`%lsmagic` lists every magic command available. Adding `?` after any of them (including a regular function) shows its docstring.

```
%lsmagic
```

```
%timeit?
```
