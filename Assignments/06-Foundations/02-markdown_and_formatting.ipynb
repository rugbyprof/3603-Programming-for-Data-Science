## 4. LaTeX Math

You can render math using LaTeX syntax inside `$ ... $` for inline or `$$ ... $$` for block math.

Example:  
Inline: $E = mc^2$  
Block:
$$
\frac{\partial f}{\partial x} = 2x
$$
