## 🔁 Count Up to 5

````
counter = 1
while counter <= 5:
    print('Count:', counter)
    counter += 1
````

## 🚫 Avoid Infinite Loops
A `while` loop keeps running as long as its condition is `True`. If you forget to update something that makes the condition eventually `False`, the loop never stops.

````
x = 0
while x < 3:
    print('x is', x)
    x += 1
````

### 🛑 Stuck in an Infinite Loop? Here's What To Do
If a `while` loop's condition never becomes `False`, the cell will run forever and the notebook will look "stuck." This happens to **everyone** at some point — it's not a disaster. In VS Code, click the **Interrupt/Stop** button next to the running cell (or use Kernel → Interrupt) to stop it, fix the bug, and run the cell again.

## ⚠️ `break` and `continue`

````
n = 0
while True:
    n += 1
    if n == 3:
        continue   # skip printing 3, go straight to the next iteration
    if n == 6:
        break      # stop the loop entirely
    print(n)
````

### 🔀 Cross-Language Note: No `do`/`while` in Python
C++ has a `do { ... } while(condition);` loop that always runs the body **at least once** before checking the condition. Python doesn't have a dedicated version of this — the common workaround is `while True:` with a `break` at the point where you'd check the condition.
```cpp
// C++
int n;
do {
    cout << "Enter a positive number: ";
    cin >> n;
} while (n <= 0);
```
```python
# Python equivalent using while True + break
while True:
    n = int(input('Enter a positive number: '))
    if n > 0:
        break
```

## 🚦 Sentinel-Controlled Loops
A common real pattern: keep looping until the user types something specific — a **sentinel value** — like `'quit'`.

````
responses = []
while True:
    word = input("Type a word (or 'quit' to stop): ")
    if word == 'quit':
        break
    responses.append(word)

print('You entered:', responses)
````
